```python
import tvscreener as tvs

# 初始化股票筛选器
ss = tvs.StockScreener()

# 获取数据（返回的是 pandas DataFrame）
df = ss.get()

# 打印前几行看看
print(f"获取到 {len(df)} 条股票数据")
print(df.head())

# 打印列名，查看都有哪些字段（如 Close, Volume, Market Cap 等）
print(df.columns)
```

```python
import tvscreener as tvs
from tvscreener import StockField

ss = tvs.StockScreener()

# --- 修正点：使用完整的字段名 ---
# 如果你想用默认的 RSI 14
ss.where(StockField.RELATIVE_STRENGTH_INDEX_14 > 70)

# 或者是当日涨幅 (CHANGE 是正确的，但有时候也叫 CHANGE_ABS 或 CHANGE_FROM_OPEN，通常 CHANGE 也是对的)
ss.where(StockField.CHANGE > 5)

df = ss.get()
print(df.head())
```

```python
from tvscreener import StockField

# 搜索包含 "RSI" 的所有字段
rsi_fields = [f for f in dir(StockField) if "RSI" in f]
print("找到的 RSI 相关字段：")
for field in rsi_fields:
    print(field)

# 搜索包含 "MACD" 的所有字段
macd_fields = [f for f in dir(StockField) if "MACD" in f]
print("\n找到的 MACD 相关字段：")
for field in macd_fields[:5]: # 只打印前5个
    print(field)
```
