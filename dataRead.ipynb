### 一、 国内市场（A股、期货、基金为主）

这些库主要针对中国金融市场，适合做量化回测和市场分析。

#### 1. Tushare Pro (最老牌、社区最成熟)

Tushare 是国内 Python 金融数据领域的鼻祖。

* **特点**：数据清洗度高，比 AKShare 更稳定（AKShare 是爬虫集合，源网站变了就需要修；Tushare 是维护自己的数据库）。
* **注意**：早期版本免费，现在的 Pro 版本实行**积分制**。基础数据免费，高频或高级数据需要积分（可以通过贡献代码或捐赠获取）。
* **官方文档**：[https://tushare.pro/document/2](https://tushare.pro/document/2)
* **源码仓库(GitHub)**：[https://github.com/waditu/tushare](https://github.com/waditu/tushare)

#### 2. Baostock (证券宝)

号称“免费、开源”的金融数据平台。

* **特点**：完全免费，不需要积分。用户注册后即可使用。数据格式对 Pandas 非常友好。主要提供历史 K 线数据，但实时性不如 AKShare。
* **官方文档**：[http://baostock.com/baostock/index.php/Python_API文档](http://baostock.com/baostock/index.php/Python_API%E6%96%87%E6%A1%A3)
* **源码仓库**：虽然它提供了 Python SDK，但它本质是客户端连接服务器，并没有公开后端数据抓取的源码，其 PyPI 仓库可见：[https://pypi.org/project/baostock/](https://pypi.org/project/baostock/)

#### 3. efinance (轻量级爬虫库)

如果你觉得 AKShare 太重了，efinance 是一个很好的替代品。

* **特点**：专门针对**东方财富网**的数据进行抓取。代码非常简洁，甚至可以直接查看源码学习如何爬取金融数据。适合只需简单的行情、资金流向数据的用户。
* **源码仓库(GitHub)**：[https://github.com/Micro-sheep/efinance](https://github.com/Micro-sheep/efinance)
* **使用文档**：直接查看 GitHub 的 Readme 即可，非常简单。

#### 4. Qstock (量化与可视化)

* **特点**：基于 Web 页面抓取，类似 AKShare，但它不仅提供数据，还集成了很多**可视化分析**（基于 Plotly）和简单的策略回测功能。
* **源码仓库(GitHub)**：[https://github.com/tkfy920/qstock](https://github.com/tkfy920/qstock)
* **说明文档**：包含在 GitHub Readme 及作者的知乎专栏中。

### 二、 国际市场（美股、外汇、全球指数）

#### 5. yfinance (Yahoo Finance)

全球最著名的开源金融数据库之一。

* **特点**：抓取 Yahoo Finance 的数据。非常适合获取美股、港股（部分）、外汇和加密货币的历史数据。无需注册 API Key，完全免费。
* **官方文档**：[https://pypi.org/project/yfinance/](https://pypi.org/project/yfinance/) (也包含在 Wiki 中)
* **源码仓库(GitHub)**：[https://github.com/ranaroussi/yfinance](https://github.com/ranaroussi/yfinance)

#### 6. Pandas-Datareader

它不是一个数据源，而是一个**接口集合**。

* **特点**：它是 Pandas 生态系统的一部分，可以统一调用 FRED (美联储数据)、World Bank (世界银行)、OECD 等宏观经济数据。
* **官方文档**：[https://pandas-datareader.readthedocs.io/](https://pandas-datareader.readthedocs.io/)
* **源码仓库(GitHub)**：[https://github.com/pydata/pandas-datareader](https://github.com/pydata/pandas-datareader)

### 三、 加密货币 (Crypto)

#### 7. CCXT

如果你涉及数字货币交易，这是唯一的“真神”。

* **特点**：支持全球 100 多个交易所的 API。不仅可以获取行情数据，还可以进行交易下单。
* **官方文档**：[https://docs.ccxt.com/](https://docs.ccxt.com/)
* **源码仓库(GitHub)**：[https://github.com/ccxt/ccxt](https://github.com/ccxt/ccxt)
