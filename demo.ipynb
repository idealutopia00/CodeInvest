```python
import yaml

config_path = r"config/config_dailyreport.yaml"

with open(config_path, 'r', encoding='utf-8') as f:
    config = yaml.safe_load(f)

print(config)
```

```python
from src.api import data_source_interface, baostock_data_source
from src.daily_report import MarketReportGenerator
import logging

# --- 核心修改：设置日志级别为 INFO ---
# format 参数用于让日志显示时间、级别和具体信息，看起来更专业
logging.basicConfig(
    level=logging.INFO, 
    format='%(asctime)s - %(name)s - %(levelname)s - %(message)s'
)

datasource : data_source_interface.FinancialDataSource = baostock_data_source.BaostockDataSource
try:
    generator = MarketReportGenerator(
        config_path=config_path,
        data_provider_class=datasource,  # <--- 这里传入你的数据源类
        save_dir="./data/reports",
        output_format="md" # 或者 "tex"
    )
    generator.run()
except Exception as e:
    print(f"❌ Error: {e}")
```
