### 行业分类

```python
import pandas as pd
import plotly.express as px
import plotly.io as pio

# === 配置 ===
pio.renderers.default = "browser"
font_family = "Microsoft YaHei"

# === 读取数据 ===
try:
    df = pd.read_excel("../Data/申万行业分类.xlsx")
except FileNotFoundError:
    print("未找到文件，使用示例数据演示")
    df = pd.DataFrame({
        "零级": ["金融","金融","金融","消费","消费"],
        "一级": ["银行","保险","券商","食品","服装"],
        "二级": ["大型银行","寿险","券商A","饮料","服饰A"],
        "三级": ["招商银行","中国平安","中信证券","康师傅","安踏"]
    })

# === Sunburst 可视化 ===
fig = px.sunburst(
    df,
    path=["零级", "一级", "二级", "三级"],  # 分层路径
    values=None,  # 可以指定权重列，如行业收入等
    title="申万行业分类层层递进可视化",
)

fig.update_layout(
    font_family=font_family,
    margin=dict(t=50, l=25, r=25, b=25)
)

fig.show()
```
