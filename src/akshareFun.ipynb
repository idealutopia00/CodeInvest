# 常用接口归档

### Notebook LL

根据您主要关注**公募基金**与**股票**的需求，我依据从**宏观经济环境**到**市场整体**，再到**行业/基金**，最后落实到**个股/具体产品**的逻辑（宏观 -> 微观），对提供的接口数据进行了重新整理和归类。

以下是按重要性层级划分的接口清单：

#### 第一层级：宏观经济与政策环境 (决定大类资产配置)
这部分数据用于判断市场所处的经济周期，决定是进攻（配置股票/偏股基金）还是防守（配置债券/货币基金）。

*   **经济增长与通胀数据（核心指标）：**
    *   `macro_china_gdp_yearly`：中国GDP年率报告，判断经济整体增速。
    *   `macro_china_cpi_monthly` / `macro_china_ppi_yearly`：CPI和PPI数据，用于分析通胀水平和企业盈利预期。
    *   `macro_china_pmi_yearly` / `macro_china_cx_pmi_yearly`：官方及财新制造业PMI，反映制造业景气度，是经济先行指标,。
*   **货币与流动性（资金面）：**
    *   `macro_china_m2_yearly`：M2货币供应年率，判断市场钱是否“充裕”。
    *   `macro_china_shrzgm`：社会融资规模增量，反映实体经济的融资需求。
    *   `macro_china_lpr`：贷款市场报价利率，直接影响资金成本。
    *   `macro_china_reserve_requirement_ratio`：存款准备金率，央行调节流动性的重要工具。
*   **汇率与外汇（影响外资流向）：**
    *   `macro_china_rmb`：人民币汇率中间价，影响出口板块及北向资金流动。
    *   `macro_china_fx_gold`：央行黄金和外汇储备。

#### 第二层级：市场整体水位与情绪 (判断入场时机)
在宏观背景下，观察目前股市的整体估值、资金流向和活跃度。

*   **大盘指数与估值：**
    *   `stock_zh_index_daily_em` / `index_zh_a_hist`：A股主要指数历史行情，观察大盘走势,。
    *   `stock_a_ttm_lyr` / `stock_a_all_pb`：全A股市场的市盈率(PE)和市净率(PB)，判断市场整体是否便宜。
    *   `stock_buffett_index_lg`：巴菲特指标，衡量市场泡沫程度。
    *   `stock_a_congestion_lg`：大盘拥挤度，预警交易过于拥挤的风险。
*   **资金流向（风向标）：**
    *   `stock_hsgt_fund_flow_summary_em`：沪深港通资金流向（北向资金），外资配置A股的重要风向标。
    *   `stock_market_fund_flow`：大盘资金流，观察主力资金的整体进出。
    *   `stock_margin_sse` / `stock_margin_szse`：融资融券汇总，反映杠杆资金的情绪,。
*   **市场情绪：**
    *   `stock_market_activity_legu`：赚钱效应分析，判断当前市场是否容易获利。
    *   `stock_zh_a_gdhs`：A股股东户数，侧面反映散户入场情况。

#### 第三层级：行业板块 (选赛道)
*   **行业指数与资金：**
    *   `stock_board_industry_hist_em` / `stock_board_industry_cons_em`：行业板块历史行情及成份股，用于筛选强势行业。
    *   `stock_sector_fund_flow_rank`：板块资金流排名，寻找资金抱团的行业。
    *   `sw_index_third_info`：申万三级行业信息，更细分的行业分类数据。

#### 第四层级：公募基金 (选产品)
基于上述宏观和行业判断，选择具体的基金产品。

*   **基金筛选与排名：**
    *   `fund_open_fund_rank_em`：开放式基金排行，筛选业绩优秀的基金。
    *   `fund_rating_all`：基金评级总汇，参考专业机构的评级。
    *   `fund_new_found_em`：新发基金数据，观察基金发行市场的热度。
    *   `fund_aum_em` / `fund_aum_trend_em`：基金公司规模排名及走势，通常规模大的公司投研更稳健,。
*   **基金经理与持仓（核心分析）：**
    *   `fund_manager_em`：基金经理大全，分析经理的过往业绩。
    *   `fund_portfolio_hold_em`：**基金持仓**，查看基金到底买了哪些股票。
    *   `fund_report_industry_allocation_cninfo`：基金行业配置，看基金经理押注了哪些赛道。
    *   `fund_portfolio_change_em`：投资组合重大变动，捕捉机构调仓动作。
*   **基金行情与交易：**
    *   `fund_open_fund_daily_em` / `fund_open_fund_info_em`：基金实时估值与历史净值。
    *   `fund_etf_spot_em`：ETF基金实时行情，适合场内交易用户。

#### 第五层级：股票个股 (选标的)
如果您直接投资股票，或者想深入研究基金重仓股，需要关注以下微观数据。

*   **个股行情与实时数据：**
    *   `stock_zh_a_spot_em` / `stock_zh_a_hist`：A股实时行情与历史日线数据,。
    *   `stock_zh_a_hist_min_em`：个股分时数据，用于日内择时。
    *   `stock_hk_spot_em` / `stock_us_spot_em`：港股与美股行情（如果是沪港通标的或QDII基金关注对象）,。
*   **基本面与财务（价值投资核心）：**
    *   `stock_financial_abstract`：财务摘要，快速了解业绩。
    *   `stock_zcfz_em` / `stock_lrb_em` / `stock_xjll_em`：资产负债表、利润表、现金流量表，深度财务分析。
    *   `stock_yjyg_em` / `stock_yjbb_em`：业绩预告与业绩报告，捕捉业绩超预期的机会,。
    *   `stock_profit_forecast_em`：机构盈利预测，看分析师对未来的预期。
*   **主力动向与筹码（技术/博弈分析）：**
    *   `stock_lhb_detail_em`：**龙虎榜详情**，查看游资和机构的买卖情况。
    *   `stock_individual_fund_flow`：个股资金流向。
    *   `stock_gdfx_holding_analyse_em`：十大股东持股分析，看是否有大基金或社保入驻。
    *   `stock_cyq_em`：筹码分布，判断获利盘和套牢盘。
*   **公司大事与消息：**
    *   `stock_news_em`：个股新闻。
    *   `stock_fhps_em`：分红配送，寻找高股息股票。
    *   `stock_restricted_release_summary_em`：限售股解禁，规避解禁抛压。
    *   `stock_jgdy_tj_em`：机构调研数据，机构频繁调研往往意味着关注度提升。

这一分类体系能够帮助您从“看天气”（宏观）到“选土壤”（行业/市场），再到“挑种子”（基金/股票），形成完整的投资决策闭环。
